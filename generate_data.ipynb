```python
import pandas as pd
import numpy as np
import os

# On passe à 2 millions de lignes pour être sûr de dépasser les 100 Mo
n_rows = 2000000
file_name = 'valeurs_immobilieres.csv'

print(f"Génération de {n_rows} lignes en cours...")

data = {
    'id_transaction': np.arange(n_rows),
    'surface_m2': np.random.uniform(20, 250, n_rows).round(2),
    'nb_pieces': np.random.randint(1, 7, n_rows),
    'distance_centre_ville': np.random.uniform(0.5, 50, n_rows).round(2),
    'annee_construction': np.random.randint(1950, 2024, n_rows),
    'indice_quartier': np.random.uniform(1, 10, n_rows).round(1),
    # On ajoute une colonne de texte pour alourdir le fichier
    'description_courte': "Zone_Residentielle_Standard_A_Verifer" 
}

df = pd.DataFrame(data)

# Logique pour le prix
df['prix_vente'] = (
    (df['surface_m2'] * 3150) + 
    (df['nb_pieces'] * 4800) - 
    (df['distance_centre_ville'] * 1100) + 
    np.random.normal(0, 15000, n_rows)
).astype(int)

print(f"Écriture du fichier {file_name}...")
df.to_csv(file_name, index=False)

size_mo = os.path.getsize(file_name) / (1024 * 1024)
print(f"Terminé ! Taille du fichier : {size_mo:.2f} Mo")
```
